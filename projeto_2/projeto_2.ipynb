####O que são estruturas de repetição ou loops

As estruturas de repetição ou loops são recursos daslinguagens de programação responsáveis por executar um bloco de código repetidamente através de determinadas condições específicas. O Python contêm dois tipos de estruturas de repetição: `for` e `While`.

####For

Usado para iterar sobre uma sequência de dados, como: listas, dicinoários, conjuntos ou executar um bloco de código para cada elemento da sequência.

####While

Também pode ser usado para iterar nestas estruturas de dados, a execução ocorre enquanto uma condição especificada for verdadeira.


####Listas (list)

Listas são coleções ordenadas e mutáveis de elementos. Os elementos podem ser de qualquer tipo, incluindo outras listas.

```python
minha_lista = [1, 2, 3, 'a', 'b']
```

```python
#Exemplo (for)
lista_nomes = ['João', 'Maria', 'Pedro']

for nome in lista_nomes:
    print(f'Olá, {nome}!')
```

```python
#Exemplo (while)
lista_nomes = ['João', 'Maria', 'Pedro']
indice = 0

while indice < len(lista_nomes):
    print(f'Olá, {lista_nomes[indice]}!')
    indice += 1
```

####Tuplas (Tuple)

Tuplas são semelhantes às listas, mas são imutáveis, o que significa que seus elementos não podem ser alterados após a criação da tupla

```python
minha_tupla = (1,2,3)
```

```python
#Exemplo For
dias_da_semana = ('Segunda', 'Terça', 'Quarta', 'Quinta', 'Sexta', 'Sábado', 'Domingo')
for dia in dias_da_semana:
    print(f'Hoje é {dia}')
```

```python
#Exemplo While
indice = 0

while indice < len(dias_da_semana):
    print(f'Hoje é {dias_da_semana[indice]}')
    indice += 1
```

`Incremento de variável` é o processo de aumentar o valor de uma variável por um determinado montante. Geralmente usado em algumas estruturas de repetição

####Conjuntos (set)

Conjuntos são coleções não ordenadas de elementos únicos. Os conjuntos são úteis para eliminar elementos duplicados e realizar operações de teoria dos conjuntos como união, interseção, etc.

```python
meu_conjunto = {1,2,3,4}
```

```python
#Exemplo for
frutas = {'maça', 'banana', 'laranja', 'maça', 'pera'}
for fruta in frutas:
    print(fruta)
```

```python
#Exemplo while
lista_frutas = list(frutas)
indice = 0

while indice < len(lista_frutas):
    print(lista_frutas[indice])
    indice += 1
```

####Dicionários (dict)

Dicionários são coleções não ordenadas de pares chave-fechadura. Eles são mutáveis e as chaves devem ser únicas e imutáveis.

```python
meu_discionario = {'nome': 'João', 'idade': 30}
```

```python
#Exemplo for
alunos_idade = {'Maria': 20, 'João': 21, 'Pedro': 22}
for aluno, idade in alunos_idade.items():
    print(f'{aluno} tem {idade} anos')
```

```python
#Exemplo while
chaves = list(alunos_idade.keys())
valores = list(alunos_idade.values())
indice = 0

while indice < len(chaves):
    aluno = chaves[indice]
    idade = valores[indice]
    print(f'{aluno} tem {idade} anos')
    indice += 1
```

####Estruturas Condicionais

Uma `Estrutura Condicinal` na programação permite que um programa execute diferentes ações com base emuma condição específica, desviando o fluxo do programa confirme necessário.


As estruturas condicionais são usadas para determinar qual bloco de código será executado a partir de uma determinada condição. O Python disponibiliza 3 comandos que ajudam a comandar o fluxo de execução, sendo: if, elif e else

```python
#if: Este é o mais básico e é utilizado para executar um bloco de código apenas se uma determinada condição for verdadeira

a = 10
b = 5

if a > b:
    print('a é maior que b')
```

```python
#elif: É utilizado para testar múltiplas condições após o if inicial. Se a condição no if inicial for falsa, o Python verifica a condição no elif

a = 10
b = 15 
c = 5

if a > b:
    print('a é maior que b')
elif a >= c:
    print('a é maior que c mas menor ou igual a b')
```

```python
#else: Este bloco será executado se nenhuma das condições anteriores (no if ou elif) for verdadeira

a = 1
b = 15 
c = 5

if a > b:
    print('a é maior que b')
elif a >= c:
    print('a é maior que 5 mas menor ou igual ao b')
else:
    print('a é menor ou igual ao b')
```

```python
#Análise de sazonalidade: Calcular a variação das vendas dos produtos por mês
total_vendas.info()
```

```python
data_format = pd.to_datetime(total_vendas['Data'], format='%m/%d/%Y')
vendas_mensais = pd.DataFrame(
    {
        'Data': data_format,
        'Quantidade': total_vendas['Quantidade']
    }
)

vendas_mensais = vendas_mensais.set_index('Data').resample('M').sum()
vendas_mensais.index = vendas_mensais.index.strftime('%b/%Y')

vendas_mensais['Variacao'] = vendas_mensais['Quantidade'].diff()
display(vendas_mensais)
```

NAN é a sigla para 'Not a Number' (não é um número). É a forma que o Pandas e outras bibliotecas no ecossistema Python usam para indicar valores faltantes ou ausentes em um conjunto de dados.

```python
#Análise de faturamento trimestral: Calcular o faturamento trimestral do ano
faturamento_trimestral = pd.DataFrame(
    {
    'Data': data_format,
    'Quantidade': total_vendas['Quantidade'],
    'Valor Unitario':  total_vendas['Valor Unitário'],
    'Faturamento': total_vendas['Quantidade'] * total_vendas['Valor Unitário']
    }
)

faturamento_trimestral['Trimestre'] = faturamento_trimestral['Data'].dt.to_period('Q')
faturamento_trimestral = pd.DataFrame(faturamento_trimestral.groupby('Trimestre')['Faturamento'].sum())

display(faturamento_trimestral)
```

`dt.to_period()`
- 'D' -> para Dias
-'M' -> para meses
- 'Y' -> para anos
- 'Q' -> para trimestres
- 'W' -> para semanas
