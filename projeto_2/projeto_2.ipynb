####O que são estruturas de repetição ou loops

As estruturas de repetição ou loops são recursos daslinguagens de programação responsáveis por executar um bloco de código repetidamente através de determinadas condições específicas. O Python contêm dois tipos de estruturas de repetição: `for` e `While`.

####For

Usado para iterar sobre uma sequência de dados, como: listas, dicinoários, conjuntos ou executar um bloco de código para cada elemento da sequência.

####While

Também pode ser usado para iterar nestas estruturas de dados, a execução ocorre enquanto uma condição especificada for verdadeira.


####Listas (list)

Listas são coleções ordenadas e mutáveis de elementos. Os elementos podem ser de qualquer tipo, incluindo outras listas.

```python
minha_lista = [1, 2, 3, 'a', 'b']
```

```python
#Exemplo (for)
lista_nomes = ['João', 'Maria', 'Pedro']

for nome in lista_nomes:
    print(f'Olá, {nome}!')
```

```python
#Exemplo (while)
lista_nomes = ['João', 'Maria', 'Pedro']
indice = 0

while indice < len(lista_nomes):
    print(f'Olá, {lista_nomes[indice]}!')
    indice += 1
```

####Dicionários (dict)

```python
#Exemplo while
chaves = list(alunos_idade.keys())
valores = list(alunos_idade.values())
indice = 0

while indice < len(chaves):
    aluno = chaves[indice]
    idade = valores[indice]
    print(f'{aluno} tem {idade} anos')
    indice += 1
```

####Estruturas Condicionais

Uma `Estrutura Condicinal` na programação permite que um programa execute diferentes ações com base emuma condição específica, desviando o fluxo do programa confirme necessário.


As estruturas condicionais são usadas para determinar qual bloco de código será executado a partir de uma determinada condição. O Python disponibiliza 3 comandos que ajudam a comandar o fluxo de execução, sendo: if, elif e else

```python
#if: Este é o mais básico e é utilizado para executar um bloco de código apenas se uma determinada condição for verdadeira

a = 10
b = 5

if a > b:
    print('a é maior que b')
```

```python
#elif: É utilizado para testar múltiplas condições após o if inicial. Se a condição no if inicial for falsa, o Python verifica a condição no elif

a = 10
b = 15 
c = 5

if a > b:
    print('a é maior que b')
elif a >= c:
    print('a é maior que c mas menor ou igual a b')
```

```python
#else: Este bloco será executado se nenhuma das condições anteriores (no if ou elif) for verdadeira

a = 1
b = 15 
c = 5

if a > b:
    print('a é maior que b')
elif a >= c:
    print('a é maior que 5 mas menor ou igual ao b')
else:
    print('a é menor ou igual ao b')
```

```python
data_format = pd.to_datetime(total_vendas['Data'], format='%m/%d/%Y')
vendas_mensais = pd.DataFrame(
    {
        'Data': data_format,
        'Quantidade': total_vendas['Quantidade']
    }
)

vendas_mensais = vendas_mensais.set_index('Data').resample('ME').sum()
vendas_mensais.index = vendas_mensais.index.strftime('%b/%Y')

vendas_mensais['Variacao'] = vendas_mensais['Quantidade'].diff()
display(vendas_mensais)
```

NaN é a sigla para 'Not a Number' (não é um número). É a forma que o Pandas e outras bibliotecas no ecossistema Python usam para indicar valores faltantes ou ausentes em um conjunto de dados.

```python
#Gráfico de linha
text_color = np.where(vendas_mensais['Variacao'] < 0, 'red', 'blue')
print(text_color)
```

```python
grafico_linha = px.line(vendas_mensais.reset_index(), x='Data', y='Quantidade', title='Vendas mensais x sazonalidadade', markers=True, text=vendas_mensais['Quantidade'])
grafico_linha.update_traces(textposition='top center', textfont=dict(color=text_color, size=15), marker=dict(color=text_color)) #bottom center, bottom left, bottom right, middle left
grafico_linha.show()
```

####Numpy

O numpy é uma biblioteca do Python utilizada para realizar operações numéricas e trabalhar com arrays multidimensionais de forma eficiente. Ele é parte fundamental do ecossistema de ciência de dados e aprendizado de máquina em Python, sendo amplamente utilizado para análise e manipulação de dados numéricos.

#####Para que serve o numpy?
- Arrays multidimensionais
O numpy fornece um objeto de array multidimensional, que é uma tabela de elementos (geralmente números), todos do mesmo tipo.

- Funções matemáticas e lógicas
O numpy fornece um grande conjunto de funções matemáticas para executar operações em arrays, incluindo operações aritméticas, lógicas, de álgebra linear, transformadas de Fourier, entre outras.

- Manipulação de dados:
Oferece métodos para manipular os dados, como ordenar, selecionar elementos únicos, etc.

- Performance:
Arrays do numpy são mais eficientes em termos de memória e desempenho comparados às listas do Python, principalmente quando lidamos com grandes volumes de dados.

- Exemplo de uso básico:

```python
#Criando um array com numpy
array = np.array([1,2,3,4,5])

#Realizando operações matemáticas
soma = np.sum(array) #soma de todos os elementos do array
media = np.mean(array) #média de todos os elementos do array

print(soma)
print(media)
```

O método `np.where()` é utilizado para fazer seleções condicionais. Ele é frequentemente usado para atribuir valores baseados em alguma condição, e sua sintaxe é: 

numpy.where(condição, valor_se_verdadeiro, valor_se_falso)

Exemplo básico:
Suponhamos que temos um array com dados de temperaturas e vamos definir que temperaturas abaixo de 20 graus são consideradas 'frio' e 20 graus ou mais são consideradas 'quente'.

```python
#Criando um array de temperaturas
temperaturas = np.array([15, 22, 18, 38, 25])

#Utilizando np.where() para atribuir 'frio' ou 'quente' de acordo com o critério
classificacao = np.where(temperaturas < 20, 'frio', 'quente')
print(classificacao)
```

```python
#Análise de faturamento trimestral: Calcular o faturamento trimestral do ano
faturamento_trimestral = pd.DataFrame(
    {
    'Data': data_format,
    'Quantidade': total_vendas['Quantidade'],
    'Valor Unitario':  total_vendas['Valor Unitário'],
    'Faturamento': total_vendas['Quantidade'] * total_vendas['Valor Unitário']
    }
)

faturamento_trimestral['Trimestre'] = faturamento_trimestral['Data'].dt.to_period('Q')
faturamento_trimestral = pd.DataFrame(faturamento_trimestral.groupby('Trimestre')['Faturamento'].sum())

display(faturamento_trimestral)
```

`dt.to_period()`
- 'D' -> para Dias
-'M' -> para meses
- 'Y' -> para anos
- 'Q' -> para trimestres
- 'W' -> para semanas

```python
text_color2 = np.where(faturamento_trimestral['Faturamento'].diff() < 0, 'red', 'blue')
#Gráfico de área
faturamento_trimestral.index = faturamento_trimestral.index.astype(str) #transformar a coluna trimestre em string para ser utilizado no gráfico 
grafico_area = px.area(faturamento_trimestral.reset_index(), x='Trimestre', y='Faturamento', title='Faturamento Trimestral', markers=True)
grafico_area.update_layout(
    xaxis_title = 'Titulo do eixo x',
    yaxis_title = 'Titulo do eixo y',
    margin = dict(t=50, l=300, r=300, b=50),#t=top, l=left, r=right, b=bottom
    title = dict(x=0.5, y=0.95) #0 -> 0%, 0.5 -> 50%, 1 -> 100%
)
grafico_area.update_traces(
    fillcolor='rgba(27, 239, 143, 0.2)',
    line_color='rgba(27, 239, 143, 1)',
    marker=dict(color=text_color2)
)
grafico_area.show()
```
