####O que são estruturas de repetição ou loops

As estruturas de repetição ou loops são recursos daslinguagens de programação responsáveis por executar um bloco de código repetidamente através de determinadas condições específicas. O Python contêm dois tipos de estruturas de repetição: `for` e `While`.

####For

Usado para iterar sobre uma sequência de dados, como: listas, dicinoários, conjuntos ou executar um bloco de código para cada elemento da sequência.

####While

Também pode ser usado para iterar nestas estruturas de dados, a execução ocorre enquanto uma condição especificada for verdadeira.


####Listas (list)

Listas são coleções ordenadas e mutáveis de elementos. Os elementos podem ser de qualquer tipo, incluindo outras listas.

```python
minha_lista = [1, 2, 3, 'a', 'b']
```

```python
#Exemplo (for)
lista_nomes = ['João', 'Maria', 'Pedro']

for nome in lista_nomes:
    print(f'Olá, {nome}!')
```

```python
#Exemplo (while)
lista_nomes = ['João', 'Maria', 'Pedro']
indice = 0

while indice < len(lista_nomes):
    print(f'Olá, {lista_nomes[indice]}!')
    indice += 1
```

####Tuplas (Tuple)

Tuplas são semelhantes às listas, mas são imutáveis, o que significa que seus elementos não podem ser alterados após a criação da tupla
